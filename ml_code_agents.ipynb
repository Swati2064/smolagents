```python
model = LiteLLMModel(
    model_id="gemini/gemini-2.5-flash",
    api_key=os.getenv("GOOGLE_API_KEY")
)
```

```python
!pip install ddgs
```

```python
ml_code_agent = CodeAgent(
    tools=[DuckDuckGoSearchTool()],
    additional_authorized_imports=[
        "pandas",
        "numpy",
        "sklearn",
        "json"
    ],
    model=model
)
```

```python
def generate_ml_code(prompt):
    full_prompt=""""
    create a step-by-step plan to write python code for svm model using sklearn to
    classify the iris dataset.
    """
    try:
        response=ml_code_agent.run(full_prompt)
        return response
    except Exception as e:
        return f"Error:{str(e)}"
```

```python
%pip install gradio
```

```python
import gradio as gr
inface=gr.Interface(
    fn=generate_ml_code,
    inputs="text",
    outputs="text",
    title="ML Code Generator ( created by Swati Jadhav)",
    description="Enter a prompt to genarate required user information"
)
```

```python
inface.launch()
```
